# An example demonstrating how to use pyGinkgo to implement sparse matrix-vector multiplication

```python
import pyGinkgo as pg

# Device initialization
executor = pg.device("cuda")

# Initialize matrix and tensors
fn = "m1.mtx"

A = pg.read(device=executor, path=fn, dtype="double", format="Csr")
n_rows = A.shape[0]

b = pg.dense(dim=(n_rows, 1), device=executor, dtype="double", fill=1.0)
x = pg.dense(dim=(n_rows, 1), device=executor, dtype="double", fill=0.0)

# Sparse Matrix Vector Product
A.apply(b, x)
```
