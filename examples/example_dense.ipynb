## PyGinkgo ```dense.matrix``` Usage Example

This notebook demonstrates basic usage of PyGinkgo dense matrices:

- Initializing and filling a dense matrix using lists
- Creating dense matrix from 1D numpy array
- Creating 1D dense matrix from numpy array with stride
- Creating dense matrix from 2D numpy array
- Creating 1D dense matrix from a .mtx file
- Scaling and inverse scaling a dense matrix

````python
import pyGinkgo as pg
import numpy as np
````

#### Initializing and filling a dense matrix using lists 

````python
ref = pg.device()

values = [[1.0, 2.0], [3.0, 4.0]]
dense = pg.matrix.dense_double(ref, values)

for i in range(2):
    print("| ", end="")
    for j in range(2):
        print(dense.at(i, j), end=" ")
    print("|")
````

#### Creating dense matrix from 1D numpy array

````python
values = [1.0, 2.0, -1.0, 3.0, 4.0, -1.0, 5.0, 6.0, -1.0, 7.4]
dense = pg.matrix.dense_double(ref, np.array(values))

for i in range(len(values)):
    print(dense.at(i))
````

#### Creating 1D dense matrix from numpy array with stride

````python
dense = pg.matrix.dense_double(ref, (3, 3), np.array([values]), 3)

for i in range(3):
    print("| ", end="")
    for j in range(3):
        print(dense.at(i, j), end=" ")
    print("|")
````

#### Creating dense matrix from 2D numpy array

````python
values = [[1.0, 2.0, -1.0, 3.0, 4.0], [-1.0, 5.0, 6.0, -1.0, 7.4]]
np_arr = np.array(values)
n_rows, n_columns = np_arr.shape
dense = pg.matrix.dense_double(ref, np.array(values))

for i in range(n_rows):
    print("| ", end="")
    for j in range(n_columns):
        print(dense.at(i, j), end=" ")
    print("|")
````

#### Creating 1D dense matrix from a .mtx file

````python
fn = "dense_example.mtx"
dense = pg.matrix.read_dense_double(fn, ref)

for i in range(12):
    print(dense.at(0, i))
````

#### Scaling and inverse scaling

````python
values = [1.0, 2.0, -1.0, 3.0, 4.0, -1.0, 5.0, 6.0, -1.0, 7.4]
dense = pg.matrix.dense_double(ref, (3, 3), np.array(values), 3)

# Scaling
dense.scale(5)
print(dense.at(1))

# Inverse scaling
dense.inv_scale(2)
print(dense.at(0))
````

````python
values = [1.0, 2.0, -1.0, 3.0, 4.0, -1.0, 5.0, 6.0, -1.0]
aT = np.array([values])
a = aT.T

dense_a = pg.matrix.dense_double(aT)
dense_b = pg.matrix.dense_double(a)
result = pg.matrix.dense_double(ref, (1, 1))

dense_a.apply(dense_b, result)
result.at(0) == aT.dot(a)[0, 0]

print(result.at(0))
````
